# 2. 函数

```python
def add_three(input_var):    # 定义简单函数
    output_var = input_var + 3
    return output_var
print(add_three(5))

new_num = add_three(10)      # 调用简单函数
print(new_num)

# 注：外部命令不能调用函数内变量，如变量output_var对外部命令来说不存在

def add_seconds(hours, minutes, seconds):           # 定义多项输入函数
    hour_min = hours * 60
    min_sec = hour_min + minutes
    total_sec = min_sec * 60 + seconds
    return total_sec
renew_seconds = add_seconds(.5, 1, 1) + 30          # 多项输入函数运算
print(renew_seconds)

def print_hello():           # 定义无输入无返回值函数
    print("Hello world!")
    print("Good morning!")
print_hello()                # 输出字符串函数
```

# 3. 数据类型

```python
# 数据类型检查：type()
# 数据长度检查：len()
# 保留小数位数：round(num, len)

# int类型数据没有长度
a = 1                        # int数据
print(a)
print(type(a))

# float类型数据没有长度
pi = 3.1415926535897932384   # float数据
almost_pi = 22/7
b = 1.
print(pi)
print(type(pi))
print(round(almost_pi, 3))   # 保留3位小数

# bool类型数据没有长度
c_one = True                 # bool数据
c_two = False
c_three = (1 < 2)
c_four = c_one
c_five = not c_one
print(c_five)
print(type(c_five))

# 注：strings类型数据运算中只能使用整数，不接受浮点数等其他类型
d = "Hello world!"           # string数据
e = d + "I love you!"
f = d * 3
g = "1.23456"
h = float(g)                 # string数据强制转换为float数据
print(g)
print(type(g))
print(len(g))
print(h)
print(type(h))
```

# 4. 条件和条件语句

```python
# 条件关系输出结果为bool数据
print(1 == 2)                # 输出1==2事件的判断结果，输出值为bool数据
# 相等：==
# 不相等：!=
# 小于：<
# 小于等于：<=
# 大于：>
# 大于等于：>=

# 条件语句
def evaluate_temp(temp):
    if temp > 38:
        message = "Fever!"
    elif temp > 35:
        message = "Normal temperature."
    else:
        message = "Low temperature."
    return message
evaluate_temp(21)
```

# 5. 列表简介

```python
# 传统记录一组信息的方式
stocks = "Nvidia, Apple, Alphabet, Microsoft, Amazon, SpaceX, Meta, Tesla, Visa, Mastercard"
print(type(stocks))
print(stocks)

# 列表记录方式
stocks_list = ["Nvidia", "Apple", "Alphabet", "Microsoft", "Amazon", "SpaceX", "Meta", "Tesla", "Visa", "Mastercard"]
print(type(stocks_list))
print(stocks_list)

# 优势：
# ①更方便获取指定位置；
# ②检查列表项目数量；
# ③添加和移除列表中项目；
print(len(stocks_list))                             # 获取列表项目数量
print("first_stock: ", stocks_list[0])              # 获取第一个列表项目，列表中从0开始计数，多项输出用逗号隔开即可
print("first three stocks: ", stocks_list[:3])      # 获取前三个列表项目
print("final two stocks: ", stocks_list[-2:])       # 获取后两个列表项目
stocks_list.remove("Meta")                          # 移除列表中的Meta项
print(stocks_list)
stocks_list.append("HSBC")                          # 添加HSBC到列表末尾位置
print(stocks_list)

# 列表中不止可以记录strings类型数据，还可以记录数字信息，并对数字信息进行加工
week_sales = [132, 154, 166, 109, 145, 198, 134]
print("Length of the list:", len(week_sales))       # 获取列表项目数量
print("Entry at index 2:", week_sales[2])           # 获取列表第2个索引位置信息
print("Minimum:", min(week_sales))                  # 列表数字最小值
print("Maximum:", max(week_sales))                  # 列表数字最大值
print("Total sales:", sum(week_sales))              # 列表所有数字加总
print("Average sales:", sum(week_sales[:7])/7)      # 列表前7个数字平均值
```

# 6. 编程入门综合练习

📝 题目名称：智能销售数据分析助手  
背景描述：你正在为一个小型网店开发一个数据处理脚本。你需要处理一周的原始销售额数据，分析其表现，并根据销售情况给出建议。  
1. 初始数据准备  
创建一个包含 6 个数字的列表 daily_sales，代表周一到周日的销售额（[120.5, 98.0, 150.2, 210.4, 88.7, 135.0]）。  
2. 定义分析函数  
数据清洗：  
发现周四的销售额（索引为 3）由于系统错误多计了 10 元，请利用减法运算对其进行修正。  
发现漏掉了一笔周日的追加订单 50.5 元，请将其添加到列表末尾。  
由于周二的订单被退货（索引为 1），请从列表中移除该项。  
统计计算：  
计算当前列表的项目数量。  
计算最高和最低销售额。  
计算这一周的总销售额。  
计算平均销售额（结果保留 2 位小数）。  
条件判断：  
如果平均销售额 大于等于 target_avg，将变量 status 设为 "达标"。  
如果平均销售额 小于 target_avg 但 大于 0，将 status 设为 "待改进"。  
如果平均销售额为 0，将 status 设为 "异常"。  
返回值：函数最后返回一个包含“总额、平均值、达标状态”的字符串信息。  
调用该函数，设置目标平均值为 130。  

打印出最终修改后的 daily_sales 列表，观察其变化。  

```python
# 数据清洗
def clear_data(basic_list):
    new_list = basic_list.copy()                        # 复制原始输入列表，并新建操作列表***
    new_list[3] = basic_list[3] - 10                    # 周四销售额多算了10
    new_list.append(50.5)                               # 周日订单50.5漏掉了
    new_list.pop(1)                                     # 周二订单98.0被退***
    return new_list

# 统计计算
def basic_data(sales_list):
    num_list = len(sales_list)                          # 当前项目数量
    min_list = min(sales_list)                          # 最低销售额
    max_list = max(sales_list)                          # 最高销售额
    sum_list = sum(sales_list)                          # 本周总销售额
    avg_list = round((sum(sales_list[:7])/7), 2)        # 本周平均销售额
    return num_list, min_list, max_list, sum_list, avg_list

# 销售状态检验
def target_status(avg_sales, target_avg):               # 输入平均销售额，目标平均销售额
    if avg_sales >= target_avg:
        result = "达标"
    elif 0 < avg_sales:
        result = "待改进"
    else:
        result = "异常"                                 # 带变量和strings类型数据输出返回值***
    return f"总额：{avg_sales*7} 平均值：{avg_sales} 达标状态：{result}"

# 总体代码执行
daily_sales = [120.5, 98.0, 150.2, 210.4, 88.7, 135.0]  # 创建一周销售额列表
clear_sales = clear_data(daily_sales)                   # 清洗后的干净数据
print(clear_sales)
basic_sales = basic_data(clear_sales)                   # 干净数据的统计信息
print(basic_sales)
status_sales = target_status(basic_sales[4], 130)       # 本周销售状态评估
print(status_sales)
print(daily_sales)                                      # 输出最初daily_sales列表
print(clear_sales)                                      # 输出最终daily sales列表
```
