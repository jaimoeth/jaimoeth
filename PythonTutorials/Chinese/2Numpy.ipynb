# 1. 导入Numpy

```python
import numpy as np              # 导入numpy库并设置别名为np

a = np.array([1, 2, 3])         # 创建简单数组[1, 2, 3]
print(a)
print(a[1])                     # 读取数组a[]中索引为1的数值 | 索引从0开始

# 数组的加减乘除运算
add_a = a + 5
sub_a = a -5
mul_a = a * 5
div_a = a / 5
print(add_a)
print(sub_a)
print(mul_a)
print(div_a)

b = np.array([                  # 创建二维数组
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
])
print(b)
print(b[1, 2])                  # 读取数组b[]中索引为[1, 2]的数值 | 先行索引，后列索引
```

# 2. 认识Numpy数组的属性

```python
import numpy as np                  # 导入numpy库并设置别名为np

a = np.array([1.25, 2, 3, 4, 5])    # 创建一维浮点数组
b = np.array([                      # 创建二维整数数组
    [1, 2, 3, 4, 5],
    [10, 20, 30, 40, 50],
    [2, 4, 6, 8, 10]
])
c = np.array(["apple", "banana"])   # 创建一维字符串数组

# 数组的属性
print(a.ndim)                       # 读取数组a[]的维度数
print(a.shape)                      # 读取数组a[]的形状(5,)代表一维数组中有5个元素
print(a.size)                       # 读取数组a[]的大小——内部元素个数
print(a.dtype)                      # 读取数组a[]的数据类型——浮点
print()
print(b.ndim)                       # 读取数组b[]的维度数
print(b.shape)                      # 读取数组b[]的形状(3, 5)代表3行5列
print(b.size)                       # 读取数组b[]的大小——内部元素个数
print(b.dtype)                      # 读取数组b[]的数据类型——整数
print()
print(c.ndim)                       # 读取数组c[]的维度数
print(c.shape)                      # 读取数组c[]的形状(2,)代表一维数组中有2个元素
print(c.size)                       # 读取数组c[]的大小——内部元素个数
print(c.dtype)                      # 读取数组c[]的数据类型——字符串
```

# 3. 创建特殊数组

```python
import numpy as np              # 导入numpy库并设置别名为np

zero = np.zeros(5)              # 创建全零数组 | 一维+五个元素
ones = np.ones((2, 2))          # 创建全一数组 | 2行2列
cont = np.arange(0, 10, 2)      # 创建连续数组 | (开始值, 结束值, 步长)其中"结束值"不输出
aver = np.linspace(0, 10, 5)    # 创建均分数组 | (开始值, 结束值, 生成数量)开始/结束值均输出
iden = np.eye(3)                # 创建单位矩阵 | 3 X 3 单位矩阵

# 打印输出
print(zero)
print(ones)
print(cont)
print(aver)
print(iden)
```

# 4. 索引与切片

```python
import numpy as np                  # 导入numpy库并设置别名为np

a = np.array([1, 2, 3, 4, 5])       # 创建一维数组
b = np.array([                      # 创建二维数组
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
])

# 对一维数组
print(a[2])                         # 读取数组a索引为2的元素
print(a[1:])                        # 读取数组a索引为1到结束的全部元素
print(a[:4])                        # 读取数组a索引从0开始到4之前的全部元素
print(a[-3:])                       # 读取数组a索引倒数第3个到最后一个全部元素
print(a[::2])                       # 读取数组a从开始到结束，步长为2的全部元素
print(a[0:4:2])                     # 读取数组a索引从0开始到4之前，步长为2的全部元素
a[2] = 234                          # 替换数组a索引为2的元素值为234
print(a)

print()

# 对二维数组
print(b[1, 2])                      # 读取数组b索引为[1, 2]的元素 | 索引1行2列元素值
print(b[1])                         # 读取数组b索引为1的整行元素
print(b[:, 2])                      # 读取数组b全部行，索引为2的整列元素
print(b[0:2, -2:])                  # 读取数组b索引0、1行，且索引-2、-1列的交集元素
b[1, 2] = 334                       # 替换数组b索引为[1, 2]的元素值为334
print(b)

# 注：切片索引的终止项均不包含！！！
```
