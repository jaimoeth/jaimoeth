# 2. Function

```python
def add_three(input_var):    # Define a simple function
    output_var = input_var + 3
    return output_var
print(add_three(5))

new_num = add_three(10)      # Calling a simple function
print(new_num)

# Note: External commands cannot access variables within a function
# For example, the variable 'output_var' does not exist for the external command

def add_seconds(hours, minutes, seconds):           # Define a multi-input function
    hour_min = hours * 60
    min_sec = hour_min + minutes
    total_sec = min_sec * 60 + seconds
    return total_sec
renew_seconds = add_seconds(.5, 1, 1) + 30          # Multi-input function operation
print(renew_seconds)

def print_hello():           # Define a function with no input and no return value
    print("Hello world!")
    print("Good morning!")
print_hello()                # Output string function
```

# 5. List Introduction

```python
# Traditional methods of recording a set of information
stocks = "Nvidia, Apple, Alphabet, Microsoft, Amazon, SpaceX, Meta, Tesla, Visa, Mastercard"
print(type(stocks))
print(stocks)

# List recording method
stocks_list = ["Nvidia", "Apple", "Alphabet", "Microsoft", "Amazon", "SpaceX", "Meta", "Tesla", "Visa", "Mastercard"]
print(type(stocks_list))
print(stocks_list)

# Advantages:
# ① It is easier to obtain a specific location;
# ② Check the number of items in the list;
# ③ Add and remove items from the list;
print(len(stocks_list))                             # Get the number of items in the list
print("first_stock: ", stocks_list[0])              
    # Get the first list item, counting from 0. Separate multiple items with commas
print("first three stocks: ", stocks_list[:3])      # Get the first three list items
print("final two stocks: ", stocks_list[-2:])       # Get the last two list items
stocks_list.remove("Meta")                          # Remove Meta items from list
print(stocks_list)
stocks_list.append("HSBC")                          # Add HSBC to the end of the list
print(stocks_list)

# Lists can record not only string data, 
# but also numeric information, and can process that numeric information.
week_sales = [132, 154, 166, 109, 145, 198, 134]
print("Length of the list:", len(week_sales))       # Get the number of items in the list
print("Entry at index 2:", week_sales[2])           # Get the second index position information of the list
print("Minimum:", min(week_sales))                  # Minimum value of numbers in the list
print("Maximum:", max(week_sales))                  # Maximum value of a number in a list
print("Total sales:", sum(week_sales))              # Sum of all numbers in the list
print("Average sales:", sum(week_sales[:7])/7)      # Average of the first 7 numbers in the list
```

# 6. Comprehensive Programming Exercises for Beginners

📝 Title: Intelligent Sales Data Analysis Assistant

Background Description: You are developing a data processing script for a small online store. You need to process a week's raw sales data, analyze its performance, and provide suggestions based on sales.

1. Initial Data Preparation

Create a list containing 6 numbers, daily_sales, representing sales from Monday to Sunday ([120.5, 98.0, 150.2, 210.4, 88.7, 135.0]).

2. Define Analysis Functions

Data Cleaning:

Found that Thursday's sales (index 3) were overstated by 10 yuan due to a system error. Please correct this using subtraction.

Found that a Sunday add-on order of 50.5 yuan was missed. Please add it to the end of the list.

Remove Tuesday's order (index 1) from the list because it was returned.

Statistical Calculations:

Calculate the number of items in the current list.

Calculate the highest and lowest sales.

Calculate the total sales for this week.

Calculate the average sales (round the result to two decimal places).

Conditional checks:

If the average sales are greater than or equal to target_avg, set the variable status to "Meets Target".

If the average sales are less than target_avg but greater than 0, set status to "Needs Improvement".

If the average sales are 0, set status to "Abnormal".

Return value: The function returns a string containing "Total, Average, and Meets Target Status".

Call this function and set the target average to 130.

Print the final modified daily_sales list and observe the changes.

```python
# Data cleaning
def clear_data(basic_list):
    new_list = basic_list.copy()                        # Copy the original input list and create a new operation list
    new_list[3] = basic_list[3] - 10                    # Thursday's sales figures were overstated by 10
    new_list.append(50.5)                               # Sunday's order of 50.5 was missed
    new_list.pop(1)                                     # Order placed on Tuesday for 98.0 was returned
    return new_list

# Statistical calculation
def basic_data(sales_list):
    num_list = len(sales_list)                          # Current number of projects
    min_list = min(sales_list)                          # Minimum sales
    max_list = max(sales_list)                          # Highest sales
    sum_list = sum(sales_list)                          # Total sales this week
    avg_list = round((sum(sales_list[:7])/7), 2)        # Average sales this week
    return num_list, min_list, max_list, sum_list, avg_list

# Sales Status Inspection
def target_status(avg_sales, target_avg):               # Enter average sales, target average sales
    if avg_sales >= target_avg:
        result = "Meets standards"
    elif 0 < avg_sales:
        result = "Needs improvement"
    else:
        result = "Abnormal"                             # Output return value with variables and string type data
    return f"Lump sum:{avg_sales*7} Average value:{avg_sales} Compliance Status:{result}"

# Overall code execution
daily_sales = [120.5, 98.0, 150.2, 210.4, 88.7, 135.0]  # Create a weekly sales list
clear_sales = clear_data(daily_sales)                   # Clean data after cleaning
print(clear_sales)
basic_sales = basic_data(clear_sales)                   # Statistical information of clean data
print(basic_sales)
status_sales = target_status(basic_sales[4], 130)       # Sales Status Assessment This Week
print(status_sales)
print(daily_sales)                                      # Output the initial daily_sales list
print(clear_sales)                                      # Output the final daily sales list
```
